```python
import numpy as np
import os
```

```python
data_dir = "./ppi-graphomer/preprocess_cpu_data"
label_file = r"./PPIdataindex.txt"

all_data = {}

# 1. merge
for i in range(16):
    file_path = os.path.join(data_dir, f"cpu{i}.npy")
    if os.path.exists(file_path):
        data = np.load(file_path, allow_pickle=True).item()
        all_data.update(data)
        print(f"Loading {file_path}, currently {len(all_data)} done")

# 2. Ouput
merged_path = os.path.join(data_dir, "final_cpu_features.npy")
np.save(merged_path, all_data)
print(f"\nSuccessfully merge {len(all_data)} files to {merged_path}")

# 3. Check
original_pdbs = set()
with open(label_file, "r") as f:
    for line in f:
        if line.strip():
            original_pdbs.add(line.split()[0].strip().lower())

missing_pdbs = original_pdbs - set(all_data.keys())
print(f"\n Totally {len(missing_pdbs)} pdbs missed")
print(missing_pdbs)
```
