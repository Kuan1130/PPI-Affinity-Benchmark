```python
import os
import shutil
from pathlib import Path
```

```python
def extract_aa_pdbs(src_dir, dest_dir, limit=None):

    src_path = Path(src_dir)
    dest_path = Path(dest_dir)

    if dest_path.exists():
        shutil.rmtree(dest_path)
    
    # Assurance
    dest_path.mkdir(parents=True, exist_ok=True)
    
    count = 0
    print(f" Scanning {src_dir}")
    
    for subdir in src_path.iterdir():
        if subdir.is_dir():
            protein_name = subdir.name
            aa_pdb_file = subdir / f"{protein_name}-aa.pdb"
            
            # -aa.pdb
            if aa_pdb_file.exists():
                # remoove -aa
                new_filename = f"{protein_name}.pdb"
                dest_file = dest_path / new_filename
                
                # duplicate
                shutil.copy2(aa_pdb_file, dest_file)
                print(f"Processing: {aa_pdb_file.name} -> {new_filename}")
                count += 1
                
                # break after limit
                if limit and count >= limit:
                    print(f"\n Reach limit: {limit}, break")
                    break
            else:
                print(f"Error: Cannot find aa in {subdir.name} ")
                
    print(f"\nSuccessfully generate {count} pdbs, saved in: {dest_dir}")
```

```python
if __name__ == "__main__":
    RAW_DATA_DIR = r"..../data/MCGLPPI_RawData/pdbs/m2_pdbbind_dimer_strict"
    
    TEST_DIR = r"./pdbs"
    
    extract_aa_pdbs(RAW_DATA_DIR, TEST_DIR, limit=999999)
```
